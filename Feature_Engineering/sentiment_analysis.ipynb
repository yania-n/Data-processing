```python
import pandas as pd
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv('reviews.csv')
```

```python
df.head()
```

Given a dataset of product reviews, with the goal of predicting the sentiment (positive or negative) of each review.

1. **TF-IDF Vectorization:** Convert raw text into a TF-IDF matrix, capturing the importance of words in context to the document corpus.
2. **Sentiment Score:** Utilize a sentiment analysis library to compute sentiment scores for each review and create a feature from these scores.
3. **N-Grams:** Create features based on the presence of specific n-grams that are hypothesized to be strong indicators of positive or negative sentiment.

Python, NLTK or similar libraries for natural language processing. For example: sklearn.feature_extraction.text using TfidfVectorizer

### step 1:

```python
vectorizer = TfidfVectorizer()
x = vectorizer.fit_transform(df['Review'])
```

```python
print(x.toarray())
x.shape
```

```python
df_data = pd.DataFrame(x.toarray(), columns= vectorizer.get_feature_names_out())
df_data.head()
```

### Step 2:

```python
from textblob import TextBlob
```

```python
df['sentiment'] = df['Review'].apply(lambda review: TextBlob(review).sentiment.polarity)
```

```python
print(df.groupby('sentiment').first())
```
