```python
# Removed all rows with no revenue value:
df_clean = df.dropna(subset= 'Revenue')
df_clean
```

```python
# Fill the 'NaN' values with 0 & remove duplicate rows:
df_clean = df_clean.fillna('0')
df_clean.drop_duplicates(inplace= True)
```

```python
# Resetting the index and discarding old index value:
df_clean.reset_index(drop=True, inplace=True)
df_clean.head()
```

```python
df_clean['Revenue'].value_counts()
```

```python
# Created a new field 'revenue_cleaned' from 'Revenue'
# Replaced 'M' with '000000' to convert the values into numeric values
# Eg. 25M -> 25000000
          
revenue_new = []                                 
for i in range(len(df_clean['Revenue'])):                 
    if (df_clean['Revenue'][i][-1] == 'M'):                            
            revenue_new.append(df_clean['Revenue'][i][:-1]+'000000')
    else:
        revenue_new.append(df_clean['Revenue'][i])

df_clean['revenue_cleaned'] = revenue_new  
df_clean.head()   
```

```python
df_clean['revenue_cleaned'].value_counts()
```

```python
# Removed '.' and '$' from the new field 'revenue_new' and casted it to type integer:
df_clean['revenue_cleaned'] = df_clean['revenue_cleaned'].str.replace('[$.]', '', regex =True)
df_clean['revenue_cleaned'] = df_clean['revenue_cleaned'].astype(int)
```

```python
# Updating column:
df_clean['Revenue'] = df_clean['revenue_cleaned']
# Droping column:
df_clean.drop(columns = ['revenue_cleaned'], inplace = True)
```

```python
# Getting the statistical values - mean, std, min, max:
stats = df_clean.describe().reset_index()
print(type(stats))
print(stats)
```

```python
# Getting the median:
median_val  = df_clean['Revenue'].median()
median_val = pd.Series(median_val)
print(type(median_val))
print(median_val)
```

```python
# Added 'median_val' as a row to the 'stats' dataframe:
stats= pd.concat([stats, median_val.to_frame()], ignore_index=True)
print(stats)
```

```python
# Renamed column 'index' as 'Stats':
stats.rename(columns= {'index':'revenue_stats'}, inplace= True)
```

```python
# Renamed column 'Revenue' as 'values':
stats.rename(columns= {'Revenue': 'values'}, inplace= True)
```

```python
# Updated the median value correctly:
stats['revenue_stats'][8] = 'median'
stats['values'][8] = stats[0][8]
stats.drop(columns= [0], inplace= True)
stats
```

```python
# Print CSV stats (mean, median, min, max, etc.,):
stats = stats.to_csv('stats_values.csv')
```

Standardizing Employee Size:

```python
# Print a specific column:
new_df = df_clean
```

```python
# Print unique values:
size_unique = new_df['Employee Size'].unique()
print(size_unique)
```

```python
new_df['Employee Size'] = new_df['Employee Size'].apply(lambda x: x.strip())
```

```python
from word2number import w2n
```

```python
# Converting alphabetic values in Employee Size to numeric values:

for i in range(len(new_df)):
    x = str(new_df.loc[i,'Employee Size'])
    
    if x.isalpha():
        new_df.loc[i,'Employee Size'] = w2n.word_to_num(new_df.loc[i,'Employee Size'])
        
    else:
        new_df.loc[i,'Employee Size'] = new_df.loc[i,'Employee Size']

new_df['Employee Size'] = new_df['Employee Size'].astype(str)
```

```python
# Print unique values:
size_unique = new_df['Employee Size'].unique()
print(size_unique)
```

```python
# Standardizing the Employee Size values into groups of -
# '0-50', '51-200', '201-500', '501-1000', '1,001-5,000', '5,001-10,000', '10,001+'

for i in range(len(new_df)):
    try:
        if new_df.loc[i,'Employee Size'].isnumeric():
            x = int(new_df.loc[i,'Employee Size'])

            if x<=50:
                new_df.loc[i,'Employee Size'] = '0-50'

            elif 51 <= x <=200:
                new_df.loc[i, 'Employee Size'] = '51-200'

            elif 201 <= x <=500:
                new_df.loc[i, 'Employee Size'] = '201-500'

            elif 501 <= x <=1000:
                new_df.loc[i, 'Employee Size'] = '501-1000'

            elif 1001 <= x <=5000:
                new_df.loc[i, 'Employee Size'] = '1,001-5,000'

            elif 5001 <= x <=10000:
                new_df.loc[i, 'Employee Size'] = '5,001-10,000'

            elif x >=10001:
                new_df.loc[i, 'Employee Size'] = '10,001+'
                
        else:
            
            if (new_df.loc[i, 'Employee Size'] == '11-50'):
                new_df.loc[i, 'Employee Size'] = '0-50'

            else:
                new_df.loc[i, 'Employee Size'] = new_df.loc[i, 'Employee Size']
                
    except ValueError:
        print("does not work")
 
new_df.reset_index(drop=True, inplace= True)
```

```python
# Print unique values:
size_unique = new_df['Employee Size'].unique()
print(size_unique)
```

# Creating a new feature 'Company type':
 Grouping companies into small, medium, and large enterprises:

```python
# Creating a new calculated field 'Company type':
# --to categorizes companies as 'Small Enterprise', 'Medium Enterprise', 'Large Enterprise'

company_type = []
for i in range(len(new_df)):
    x = new_df.loc[i, 'Employee Size']
    y = new_df.loc[i, 'Revenue'] 
    if ((y < 38500000)&((x == '0-50')|(x == '51-200')|(x == '201-500')|(x == '501-1000'))):
        c_type = 'Small Enterprise'
        company_type.append(c_type)

    elif ((38500000 <= y < 1000000000)&((x == '1,001-5,000')|(x == '5,001-10,000'))):
        c_type = 'Medium Enterprise'
        company_type.append(c_type)

    elif ((y >= 1000000000)&(x == '10,001+')):
        c_type = 'Large Enterprise'
        company_type.append(c_type)
    else:
        c_type = ''
        company_type.append(c_type)
 
new_df['Company type'] = company_type
```
