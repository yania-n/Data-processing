```python
import pandas as pd

# The dataset has two features: 'area' and 'price'
data = {'area': [1000, 1500, 2000, 2500, 3000],
        'price': [500000, 600000, 700000, 800000, 900000]}
df = pd.DataFrame(data)
print("Original Dataset")
print(df)
print("----------------")


# We can create a new feature: 'price per square meter'
df['price_per_meter'] = df['price'] / df['area']

# Now we have a new feature that might be more useful for a machine learning model
print("Feature Dataset")
print(df)
print("----------------")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Let's assume we have a simple dataset of customer reviews
# The dataset has one feature: 'review'
data = {'review': [
    "I love this product, must try", 
    "This product is great", 
    "I hate this product it's terrible", 
    "This product is terrible, don't recommend"
]}
df = pd.DataFrame(data)

# We can create a new feature: 'word count'
df['word_count'] = df['review'].apply(lambda x: len(str(x).split()))

# We can also create a new feature: 'sentiment'
# For simplicity, let's assume that 'love' and 'great' are positive words, and 'hate' and 'terrible' are negative words
df['sentiment'] = df['review'].apply(lambda x: 1 if ('love' in x or 'great' in x) else (-1 if ('hate' in x or 'terrible' in x) else 0))

# Now we have new features that might be more useful for a machine learning model
print("New Features")
print(df)
print("------------------------------------------")

# We can also use CountVectorizer to convert the text data into a matrix of token counts
# CountVectorizer is a text preprocessing technique commonly used in natural language 
# processing (NLP) tasks for converting a collection of text documents into a numerical representation
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['review'])

# This can be used as input for a machine learning model
print("Matrix of tokens Features")
print(X.toarray())
print("------------------------------------------")

```
