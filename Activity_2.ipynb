```python
import pandas as pd
```

```python
df = pd.read_csv('activity1_companies.csv')
```

```python
df.head()
```

```python
df['Employee Size'].unique()
```

```python
# Converting alphabetic values in Employee Size to numeric values:
from word2number import w2n
for i in range(len(df)):
    x = str(df.loc[i,'Employee Size'])
    
    if x.isalpha():
        df.loc[i,'Employee Size'] = w2n.word_to_num(df.loc[i,'Employee Size'])
        
    else:
        df.loc[i,'Employee Size'] = df.loc[i,'Employee Size']

df['Employee Size'] = df['Employee Size'].astype(str)
```

```python
df['Employee Size'].unique()
```

```python
# df[df['Employee Size'].str.contains('-')]
df['Employee Size'] = df['Employee Size'].str.replace(',', '')
for i in range(len(df)):
    if '-' in df.loc[i, 'Employee Size']:
        num1= float(df.loc[i,'Employee Size'].split('-')[0])
        num2= float(df.loc[i,'Employee Size'].split('-')[1])

        Emp_avg = int((num1+num2)/2)

        df.loc[i,'Employee Size'] = str(Emp_avg)
    else:
        df.loc[i,'Employee Size'] = df.loc[i,'Employee Size']


```

```python
df['Employee Size'].unique()
```

```python
print(df['Revenue'])
```

```python
revenue_new = []                                 
for i in range(len(df['Revenue'])):     
    if (df.loc[i, 'Revenue'][-1] == 'M'):

        if '.' in df.loc[i, 'Revenue']:
            revenue_new.append(df.loc[i, 'Revenue'][:-1]+'00000')

        else:                        
            revenue_new.append(df.loc[i, 'Revenue'][:-1]+'000000')

    else:
        revenue_new.append(df.loc[i, 'Revenue'])

df['revenue_cleaned'] = revenue_new  

df['revenue_cleaned'] = df['revenue_cleaned'].str.replace('[$.]', '', regex =True)
df['revenue_cleaned'] = df['revenue_cleaned'].astype(int)

# Updating column:
df['Revenue'] = df['revenue_cleaned']
df.drop(columns = ['revenue_cleaned'], inplace = True)
```
