```python
# df[df['Employee Size'].str.contains('-')]
num1 = df['Employee Size'].str.split('-').str[0]
num2 = df['Employee Size'].str.split('-').str[1]
for i in range(len(df['Employee Size'])):
    if df['Employee Size'][i].str.contains('-'):
        x = num1[i]
        y = num2[i]
        df['Employee Size'][i] = (x+y)/2
    else:
        df['Employee Size'][i] = num1[i]
```
