```python
# Fill all 'NaN' with 0:
df_clean = df.fillna(0)
df_clean.drop_duplicates(inplace= True)
```

```python
df_clean.reset_index(inplace=True)
df_clean.drop(columns=['index'],inplace=True)
df_clean.head()
```

```python
df_clean['Revenue'] = df_clean['Revenue'].astype(str)
```

```python
# Transformed 'Revenue' field to a numeric format and created a new field 'revenue_cleaned'
          
revenue_new = []                                 
for i in range(len(df_clean['Revenue'])):                 
    if (df_clean['Revenue'][i] == '0'):
        revenue_new.append(df_clean['Revenue'][i])
    else:
        if (df_clean['Revenue'][i][-1] == 'M'):                            
            revenue_new.append(df_clean['Revenue'][i][:-1]+'000000')
        else:
            revenue_new.append(df_clean['Revenue'][i])
print(revenue_new)
       
```

```python
df_clean['revenue_cleaned'] = revenue_new
df_clean.head(3)
```

```python
pd.set_option('display.max_row', None)
df_clean['revenue_cleaned'].value_counts()
```

```python
# Removing '.' and '$' from the new field 'revenue_new' and casting it as integer type:
df_clean['revenue_cleaned'] = df_clean['revenue_cleaned'].str.replace('[$.]', '', regex =True)
df_clean['revenue_cleaned'] = df_clean['revenue_cleaned'].astype(int)
```

```python
# Getting statistical values into the variable 'stats':
stats = df_clean.describe().reset_index()
print(type(stats))
print(stats)
```

```python
# Getting the median value for revenue:
median_val  = df_clean['revenue_cleaned'].median()
median_val = pd.Series(median_val)
print(type(median_val))
print(median_val)
```

```python
# Concateneting 'median_val' to the 'stats' dataframe:
stats= pd.concat([stats, median_val.to_frame()], ignore_index=True)
print(stats)
```

```python
# Renaming column 'index' as 'Stats':
stats.rename(columns= {'index':'revenue_stats'}, inplace= True)
```

```python
stats.rename(columns= {'revenue_cleaned': 'values'}, inplace= True)
```

```python
# Updating the median value correctly
stats['revenue_stats'][8] = 'median'
stats['values'][8] = stats[0][8]
stats.drop(columns= [0], inplace= True)
```
