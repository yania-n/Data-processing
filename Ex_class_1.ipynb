```python
def convert_number_word_to_int(word):
    number_dict = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5, 'six': 6, 'seven': 7, 'eight': 8, 'nine': 9, 'ten': 10, 'eleven': 11, 'twelve': 12}
    return number_dict.get(word.lower(), word)

clean = []
```

```python
for i in rawData:
    if type(i)== str:
        x = convert_number_word_to_int(i)

    else:
        x = i
    
    clean.append(int(x))

print(clean)
```

```python
for i in rawData:
    try:
        x = int(i)
        clean.append(x)
    except ValueError:
        x= convert_number_word_to_int(i)
        clean.append(x)
print(clean)


```
